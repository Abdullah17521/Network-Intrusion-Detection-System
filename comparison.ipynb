```python
import matplotlib.pyplot as plt
import numpy as np

# Teeno models ki metrics ka data
models = ['Naive Bayes', 'KNN', 'K-Means (Clustering)']
accuracy = [81.4, 93.2, 0.0]       # K-Means ki classification accuracy direct nahi hoti
recall = [88.5, 91.0, 0.0]         # K-Means ka recall nahi hota
silhouette = [0.0, 0.0, 42.0]      # Silhouette score ko 100 ki scale par dikhane ke liye (0.42 * 100)

x = np.arange(len(models))  
width = 0.25  # Bars ki width thodi kam ki taake teeno fit aa skein

fig, ax = plt.subplots(figsize=(10, 6))

# Bars plotting
rects1 = ax.bar(x - width, accuracy, width, label='Classification Accuracy %', color='#00FFCC')
rects2 = ax.bar(x, recall, width, label='Recall % (Attack Catch Rate)', color='#FF0033')
rects3 = ax.bar(x + width, silhouette, width, label='Cluster Density (Silhouette x 100) %', color='#3399FF')

# Formatting Labels and Titles
ax.set_ylabel('Performance Percentage (%)', fontsize=12)
ax.set_title('Complete Project Showdown: Supervised vs Unsupervised Models', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(models, fontsize=11)
ax.set_ylim(0, 110)
ax.legend(loc='upper right')
ax.grid(axis='y', linestyle='--', alpha=0.5)

# Numbers ko bars ke ooper display karne ka function
def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
        if height > 0:  # Sirf un bars par likhein jahan data majood hai
            ax.annotate(f'{height}%' if rects != rects3 else f'{height/100}',
                        xy=(rect.get_x() + rect.get_width() / 2, height),
                        xytext=(0, 3),  
                        textcoords="offset points",
                        ha='center', va='bottom', fontsize=10)

autolabel(rects1)
autolabel(rects2)
autolabel(rects3)

plt.tight_layout()
plt.show()
```
